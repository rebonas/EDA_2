## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

#verify if the type_of_loan matches with the number of loans
def match_count(x):
  if x=="No Data":
    return 0
  loans=x.replace(' and ', ', ').split(',')
  loans=[p.strip() for p in loans if p.strip() != '']
  return len(loans)
data['Loan_Count_from_Type'] = data['Type_of_Loan'].apply(match_count)
data['Match_Count']=data['Num_of_Loan'] == data['Loan_Count_from_Type']
print(data['Match_Count'].value_counts())


````

#### Chart - 9

````python
# Chart - 9 visualization code
order = ['Poor', 'Standard', 'Good']
col = 'Interest_Rate'

def count_outliers(s):
    q1, q3 = s.quantile([0.25, 0.75])
    iqr = q3 - q1
    return ((s < q1 - 1.5*iqr) | (s > q3 + 1.5*iqr)).sum()

counts = data.groupby('Credit_Score')[col].apply(count_outliers)
print(counts)

plt.figure(figsize=(8,5))
ax = sns.boxplot(data=data, x='Credit_Score', y=col,
                 order=order, hue='Credit_Score', hue_order=order,
                 palette='viridis', legend=False)


ax.set_xticks(range(len(order)))
ax.set_xticklabels([f'{s}\n(outliers: {counts[s]:,})' for s in order])

plt.title(f'{col} across Credit Score Categories')
plt.xlabel('Credit Score')
plt.ylabel(col)
plt.show()
````

##### 1. Why did you pick the specific chart?

A box plot compares the median,spread and outliers of each credit score and interest rate group side by side.
